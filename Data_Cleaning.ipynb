```python
import pandas as pd

# Step 1: Load CSV
df = pd.read_csv('/content/miami_air_quality_2021_2024.csv')

# Step 2: Print original shape and raw column names
print("Original shape:", df.shape)
print("Original columns:", df.columns.tolist())

# Step 3: Clean column names (strip whitespace)
df.columns = df.columns.str.strip()

# Step 4: Drop all-null rows
df = df.dropna(how='all')

# Step 5: Drop rows with missing AQI values (Arithmetic Mean)
if 'Arithmetic Mean' in df.columns:
    df = df.dropna(subset=['Arithmetic Mean'], how='any')
else:
    raise KeyError("Expected column 'Arithmetic Mean' not found after stripping. Found columns: {}".format(df.columns.tolist()))

# Step 6: Drop duplicate rows
df = df.drop_duplicates()

# Step 7: Show cleaned shape and preview
print("Cleaned shape:", df.shape)
df.head()
```

```python
output_excel_path = "/content/cleaned_miami_air_quality_2021-24.xlsx"
df.to_excel(output_excel_path, index=False)
print(f"Excel saved to: {output_excel_path}")
```
