```python
import pandas as pd

# Step 1: Load CSV (treat empty strings and spaces as NaN)
df = pd.read_csv('Illionois_air_quality_2021_2024.csv', na_values=["", " "])

# Step 2: Print original shape and column names
print("Original shape:", df.shape)
print("Original columns:", df.columns.tolist())

# Step 3: Clean column names
df.columns = df.columns.str.strip()

# Step 4: Drop rows where all columns are NaN
df = df.dropna(how='all')

# Step 5: Drop rows with missing or zero 'AQI'
required_cols = ['AQI', 'Parameter Name', 'City Name']
for col in required_cols:
    if col not in df.columns:
        raise KeyError(f"Expected column '{col}' not found. Found columns: {df.columns.tolist()}")

# Remove missing or zero AQI values
df = df[df['AQI'].notna()]         # Remove NaNs
df = df[df['AQI'] != 0]            # Remove zero AQI values
df.head()
```

```python

# Step 6: Filter for Parameter = PM2.5
df = df[df['parameter'].str.strip().str.lower() == 'pm25']

# Step 7: Filter for selected cities
allowed_cities = ['New York', 'Chicago', 'Los Angeles', 'Miami']
df = df[df['City Name'].str.strip().isin(allowed_cities)]

# Step 8: Drop duplicate rows
df = df.drop_duplicates()

# Step 9: Print cleaned shape and preview
print("Cleaned shape:", df.shape)
print(df.head())
```

```python

# Step 10: Save to Excel
output_excel_path = "cleaned_Illionois_air_quality_2021_2024.xlsx"
df.to_excel(output_excel_path, index=False)
print(f"Excel saved to: {output_excel_path}")
```
