```python
import requests
import zipfile
import io
import pandas as pd

# EPA Parameter codes
parameters = {
    "pm25": 88101,
    "pm10": 81102,
    "co": 42101,
    "so2": 42401,
    "no2": 42602,
    "o3": 44201,
}

# Florida - Miami-Dade County
state_code = 12   # Florida
county_code = 86  # Miami-Dade County

start_year = 2021
end_year = 2024

def download_epa_data(year, parameter_code):
    url = f"https://aqs.epa.gov/aqsweb/airdata/daily_{parameter_code}_{year}.zip"
    print(f"Downloading {url}")
    response = requests.get(url)
    if response.status_code == 200:
        with zipfile.ZipFile(io.BytesIO(response.content)) as z:
            csv_filename = z.namelist()[0]
            with z.open(csv_filename) as f:
                df = pd.read_csv(f)
                return df
    else:
        print(f"Failed to download data for {year} and parameter {parameter_code}")
        return None

all_dfs = []

for year in range(start_year, end_year + 1):
    for param_name, param_code in parameters.items():
        df = download_epa_data(year, param_code)
        if df is not None:
            # Filter for Florida Miami-Dade County
            df_filtered = df[(df['State Code'] == state_code) & (df['County Code'] == county_code)].copy()
            if not df_filtered.empty:
                df_filtered['parameter'] = param_name
                all_dfs.append(df_filtered)

# Combine all pollutant data into one DataFrame
combined_df = pd.concat(all_dfs, ignore_index=True)

# Save to CSV
combined_df.to_csv("miami_air_quality_2021_2024.csv", index=False)
print(f"Saved combined data with {len(combined_df)} rows to miami_air_quality_2020_2024.csv")
```
